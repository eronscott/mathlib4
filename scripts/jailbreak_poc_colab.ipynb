## 1) 모듈 로드
리포지토리 루트 기준으로 스크립트를 import합니다.


```python
import importlib.util
from pathlib import Path

module_path = Path("scripts/jailbreak_poc_colab.py")
spec = importlib.util.spec_from_file_location("jbpoc", module_path)
jbpoc = importlib.util.module_from_spec(spec)
spec.loader.exec_module(jbpoc)
```

## 2) Mock 모드 실행 (API 키 불필요)


```python
result = jbpoc.run_experiment(
    use_openai=False,
    emb_model="text-embedding-3-small",
    judge_model="gpt-4.1-mini",
    seed=7,
)

print("-- Spectral gap (lambda1) --")
print(f"{result.lambda1:.6f}\n")
print("-- Meaning manifold curvature --")
print(f"{result.curvature:.6f}\n")
print("-- Residual attack energy (epsilon) --")
print(f"{result.epsilon:.6f}\n")
print("-- Adaptive trajectory (first 12) --")
for row in result.adaptive_trajectory:
    print(row)
```
